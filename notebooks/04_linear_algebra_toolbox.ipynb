## Before you start

A matrix can stretch, rotate, project, or mix numbers. Linear algebra gives tools to understand or reverse these changes. `solve` finds x in Ax=b. `eigh` finds orthogonal directions for a symmetric matrix. SVD factors a rectangular matrix into directions and nonnegative strengths.

NumPy returns **Vt**, not V. `(U*s) @ Vt` reconstructs W. For S=W W.T, eigenvalues are s², and its squared Frobenius norm is the sum of s⁴. Keep the largest strengths for a compact approximation.
