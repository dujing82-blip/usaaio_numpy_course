## Before you start

A dot product multiplies matching entries and adds the products. Matrix multiplication repeats that calculation for every row/column pair. `(M,N) @ (N,P)` has shape `(M,P)`: the shared N is summed away. `*` never performs that sum.

Samples are usually rows of X. `X @ theta` predicts once per sample. `X.T @ r` weights all sample rows and adds them to produce one number per feature. Both can be rewritten using broadcasting and sum.
