## Before you start

Elementwise means doing the same calculation at each address. Broadcasting lets a smaller array act on many entries without making explicit copies. Compare shapes from the **right**: each pair must be equal, or one must be 1. Missing leading dimensions act like 1.

| Shapes | Result |
|---|---|
| (2,3) and (3,) | (2,3) |
| (2,3) and (2,1) | (2,3) |
| (2,3) and (2,) | error |
| (4,1) and (1,3) | (4,3) |

A reduction combines many numbers: `sum(axis=1)` adds across columns, leaving one result per row. `keepdims=True` keeps the reduced axis with length one, making later broadcasting easier.

**Task:** `softmax(z)`

**Rules:** softmax: no-loops

**Predict first:** What shape must the denominator have to divide each row?

**My prediction:** _Write your answer here before coding._

```python
def softmax(z):
    'softmax: input z: (...,K); output: (...,K); last-axis probabilities sum to 1. See the exercise task for the required behavior.'
    e = np.exp(z-z.max(axis=-1,keepdims=True))
    return e/e.sum(axis=-1,keepdims=True)
```

### Exercise 2.3 — Broadcast or error? ★★

[B2]

**Source:** Operation practice adapted from the supplied knowledge map. Adapted exercise; source tags supplied with the course specification.

**Task:** `broadcast_result(shape_a, shape_b)` → resulting shape tuple or `None` (student implements the right-alignment rule by hand, no NumPy allowed inside; grader also checks against `np.broadcast_shapes`)

**Shape contract:** input: two shape tuples; output: shape tuple or None

**Rules:** broadcast_result: no-numpy

**Predict first:** Before running, write the input and output shapes for each function. Contract: input: two shape tuples; output: shape tuple or None.

Loops are allowed here because you are implementing the shape rule, not a numerical array calculation.

**My prediction:** _Write your answer here before coding._

```python
def broadcast_result(shape_a,shape_b):
    'broadcast_result: input: two shape tuples; output: shape tuple or None. See the exercise task for the required behavior.'
    out = []
    a,b = list(shape_a),list(shape_b)
    while a or b:
        x = a.pop() if a else 1
        y = b.pop() if b else 1
        if x != y and x != 1 and y != 1:
            return None
        out.append(y if x == 1 else x)
    return tuple(reversed(out))
```
